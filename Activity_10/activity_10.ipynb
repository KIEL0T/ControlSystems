```python
from openai import OpenAI
from langchain_community.chat_message_histories import ChatMessageHistory
import keyboard

# Connect to LM Studio local server
client = OpenAI(
    base_url="http://localhost:1234/v1",
    api_key="lm-studio"  # any string works
)
```

```python
# Memory
history = ChatMessageHistory()

system_message = {
    "role": "system",
    "content": (
        "You are a helpful assistant. "
        "You remember details the user shares (like their name, preferences, etc.) "
        "and use them in future responses."
    )
}
```

```python
while True:
    query = input("Ask a question || Make a statement")

    # Exit condition
    if keyboard.is_pressed('Esc'):
        print("\nGoodbye. See you later!")
        break

    if not query.strip():
        continue

    print("User: " + query)

    # Save message to history
    history.add_user_message(query)

    # Build messages list (START with system message)
    messages_with_history = [system_message]

    # Convert LangChain to OpenAI format
    for messages in history.messages:
        if messages.type == "human":
            role = "user"
        else:
            role = "assistant"
            
        messages_with_history.append({
            "role": role,
            "content": messages.content
        })

    # Responses
    responses = client.chat.completions.create(
        model="qwen/qwen3-4b-2507",
        messages=messages_with_history,
        temperature=0
    )

    ai_reply = responses.choices[0].message.content.strip()
    
    # Save AI response
    history.add_ai_message(ai_reply)
    # Format AI response
    print(f'AI-ssistant: {ai_reply}')
    print("-" * 200)
```
