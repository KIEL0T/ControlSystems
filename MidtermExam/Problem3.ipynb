```python
import control as ct
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Define the transfer function G(s) = num / den
# G(s) = 3 / (8s^4 + 12s^3 + 26s^2 + 18s)
num = [3]
den = [8, 12, 26, 18, 0] # Coefficients of the denominator polynomial

system = ct.TransferFunction(num, den)
```

```python
# Compute Poles and Zeros
poles, zeros = ct.pzmap(system, plot=False) # plot=False to get values

print("--- System Analysis Results ---")
print("\nComputed Poles:")
for pole in poles:
    print(f"- {pole:.4f}") # Format to 4 decimal places

print("\nComputed Zeros:")
if zeros.size == 0:
    print("- None (The system has no zeros).")
else:
    for zero in zeros:
        print(f"- {zero:.4f}")
```

```python
# Step Response
plt.figure()
time, response = ct.step_response(system)
plt.plot(time, response)
plt.title('Step Response')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.grid(True, linestyle='--', linewidth=0.5)
plt.show()
```

```python
# Plotting (as shown in the provided images)
# Pole-Zero Map
plt.figure()
ct.pzmap(system, grid=True, title='Pole-Zero Map')
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.axhline(0, color='gray', linewidth=0.8)
plt.axvline(0, color='gray', linewidth=0.8)
plt.show()
```
