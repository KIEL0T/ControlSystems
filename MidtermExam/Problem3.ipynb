```python
import control
import matplotlib.pyplot as plt
```

```python
# Define Transfer Function H(s) = 3 / (8s^4 + 12s^3 + 26s^2 + 18s)
num = [3]
den = [8, 12, 26, 18, 0]
system = control.TransferFunction(num, den)
```

```python
# Calculate System Dynamics
poles = control.poles(system)
zeros = control.zeros(system)
print(f"Poles: {poles}")
print(f"Zeros: {zeros}")
```

```python
# Visualization 1: Time Domain Response
t, y = control.step_response(system)
plt.figure(figsize=(10, 7))
plt.plot(t, y, color='#2563eb', linewidth=2.5)
plt.title("Step Response (Integrating System)")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
# Visualization 2: Pole-Zero Map
plt.figure(figsize=(10, 7))
control.pzmap(system, grid=True)
plt.title("System Pole-Zero Map Analysis")
plt.grid(True, linestyle='--', alpha=0.6)
```
