```python
import control as ctrl
import matplotlib.pyplot as plt
```

```python
# Numerator and denominator coefficients extracted from derivation
num = [3]
den = [8, 12, 26, 18, 0]

# Create the transfer function G = ctrl.tf(num, den)
G = ctrl.tf(num, den)

# Print the transfer function
print("System Transfer Function G(s):")
print(G)

```

```python
# Compute and print poles and zeros
poles = ctrl.poles(G)
zeros = ctrl.zeros(G)
```

```python
# Generate a pole-zero map
plt.figure(figsize=(8, 6))
ctrl.pzmap(G, grid=True)
plt.title('Pole-Zero Map')
```

```python
# Plot the system step response
plt.figure(figsize=(8, 6))
t, y = ctrl.step_response(G)
plt.plot(t, y)
plt.grid(True)
plt.show()
```
