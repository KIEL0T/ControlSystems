```python
import control
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define system coefficients
num = [1, 1]
den = [1, 4, 7, 6, 3]
```

```python
# Create transfer function H(s)
system = control.TransferFunction(num, den)
```

```python
# Calculate poles and zeros
poles = control.poles(system)
zeros = control.zeros(system)

print(f"Poles: {poles}")
print(f"Zeros: {zeros}")
```

```python
# Generate Step Response
t, y = control.step_response(system)
plt.figure(figsize=(8, 6))
plt.plot(t, y, linewidth=2, color='#2563eb')
plt.title('System Step Response')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.grid(True)
plt.show()
```

```python
# Generate Pole-Zero Map
plt.figure(figsize=(8, 6))
control.pzmap(system, grid=True)
plt.title('Pole-Zero Map')
plt.show()
```
