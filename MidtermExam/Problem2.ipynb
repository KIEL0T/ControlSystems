```python
import control as ctrl
import matplotlib.pyplot as plt
```

```python
# Define numerator and denominator coefficients extracted from the derivation
num = [1]
den = [1, 2, 3, 2, 0]
```

```python
# Create the transfer function G = ctrl.tf(num, den)
G = ctrl.tf(num, den)

# Print the transfer function
print("System Transfer Function G(s):")
print(G)
```

```python
# Compute and print poles and zeros
poles = ctrl.poles(G)
zeros = ctrl.zeros(G)
print("\nPoles of the system:", poles)
print("Zeros of the system:", zeros)
```

```python
# Generate a pole-zero map using ctrl.pzmap
plt.figure()
ctrl.pzmap(G, grid=True)
plt.title('Pole-Zero Map')
```

```python
# Plot the system step response
plt.figure()
t, y = ctrl.step_response(G)
plt.plot(t, y)
plt.title('System Step Response')
plt.xlabel('Time')
plt.ylabel('Amplitude')
plt.grid(True)

# Display the plots
plt.show()
```
