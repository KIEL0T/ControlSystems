```python
import numpy as np
import matplotlib.pyplot as plt
import control as ct
```

```python
# Define the control system based on the provided Laplace equations:
# Equation 1: T(s) = θ1(s)[s^2 + s + 1] - θ2(s)
# Equation 2: θ1(s) = θ2(s)[s^2 + s + 1]
#
# Substituting θ1(s) from Eq 2 into Eq 1:
# T(s) = θ2(s)[s^2 + s + 1]^2 - θ2(s)
# T(s) = θ2(s)[(s^2 + s + 1)^2 - 1]
# Expanding (s^2 + s + 1)^2: s^4 + 2s^3 + 3s^2 + 2s + 1
# T(s) = θ2(s)[s^4 + 2s^3 + 3s^2 + 2s]
#
# Target Transfer Function G(s) = θ2(s) / T(s):
# G(s) = 1 / (s^4 + 2s^3 + 3s^2 + 2s)

# Numerator and Denominator coefficients
num = [1]
den = [1, 2, 3, 2, 0]

# Define the system transfer function
system = ct.TransferFunction(num, den)
```

```python
# 1. Calculate the system's poles and zeros
poles = ct.poles(system)
zeros = ct.zeros(system)

print("System Model: G(s) = θ2(s)/T(s)")
print(system)
print("-" * 30)
print(f"Poles: {poles}")
print(f"Zeros: {zeros}")
print("-" * 30)
```

```python
# 2. Generate plots
plt.figure(figsize=(8, 6))

# Plot 1: Pole-Zero Map
ct.pzmap(system, grid=True)
plt.title('Pole-Zero Map')
```

```python
# Plot 2: Step Response
plt.subplot(1, 2, 2)
t, y = ct.step_response(system)
plt.plot(t, y)
plt.title('Step Response Graph')
plt.xlabel('Time (seconds)')
plt.ylabel('Theta 2 (Amplitude)')
plt.grid(True)

plt.tight_layout()
plt.show()
```
