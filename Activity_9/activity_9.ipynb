```python
import sympy as sp
import control as ctrl
import matplotlib.pyplot as plt

# 1. Define Symbols
s = sp.symbols('s')
X = sp.symbols('X')
Y = sp.symbols('Y')
```

```python
# 2. Represent the Equation: 2*s^2*Y + s*Y = s*X + X
# Replace'D' with 's' in accordance to Laplace Transform
A = 2*s**2 * Y + s * Y
B = s * X + X
```

```python
# 3. Solve for Y/X (Transfer Function)
# Mathematical Equation: Y(2s^2 + s) = X(s + 1) -> Y/X = (s + 1) / (2s^2 + s)
equation_tf = sp.solve(sp.Eq(A, B), Y)[0] / X
simplified_tf = sp.simplify(equation_tf)

print(f"Transfer Function G(s): {simplified_tf}")
```

```python
# 4. Convert to Control Library Format
num_coeffs = [float(c) for c in sp.Poly(sp.numer(simplified_tf), s).all_coeffs()]
den_coeffs = [float(c) for c in sp.Poly(sp.denom(simplified_tf), s).all_coeffs()]

sys = ctrl.tf(num_coeffs, den_coeffs)
print(sys)
```

```python
# 5. Poles and zeros output value
print(f"Poles: {ctrl.poles(sys)}")
print(f"Zeros: {ctrl.zeros(sys)}")
```

```python
# 6. Poles and Zeros
plt.figure()
ctrl.pzmap(sys, grid=True)
plt.title('Pole-Zero Map')
plt.show()
```

```python
# 7. Plot Step Response
t, y = ctrl.step_response(sys)
plt.plot(t, y)
plt.title('Step Response')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.grid(True)
plt.show()
```
